```python
class Node:
    def __init__(self,val):  # Create node
        self.val=val
        self.next=None
```

```python
# ---------------- OBJECT ----------------
head = Node(10)
second = Node(20)
third = Node(30)
fourth = Node(40)

head.next = second
second.next = third
third.next = fourth
```

```python
print("************** ORIGINAL LINKED LIST **************")
curr = head
while curr:
    print(curr.val, end=" -> ")
    curr = curr.next
print("None")
```

```python
# --------------------Insert in linkedlist----------------------
def insert_head(head,new_node_value):   # Insert at head
    curr=head
    print("\n------ BEFORE INSERT AT HEAD ------")
    while curr!=None:
        print(curr.val,end=" -> ")
        curr=curr.next
    print("None")

    new_node=Node(new_node_value)
    new_node.next=head
    head=new_node

    curr=head
    print("##### AFTER INSERT AT HEAD #####")
    while curr!=None:
        print(curr.val,end=" -> ")
        curr=curr.next
    print("None")
    return head
head = insert_head(head,5)
```

```python
def insert_end(new_node_value):    # Insert at end
    global head
    curr=head

    print("\n************** BEFORE INSERT AT END **************")
    while curr!=None:
        print(curr.val,end=" -> ")
        curr=curr.next
    print("None")

    new_node=Node(new_node_value)
    curr=head

    while curr.next!=None:
        curr=curr.next

    curr.next=new_node
    new_node.next=None

    curr=head
    print("------ AFTER INSERT AT END ------")
    while curr!=None:
        print(curr.val,end=" -> ")
        curr=curr.next
    print("None")
insert_end(50)
```

```python
def insert_at_place(new_node_value,k):
    global head
    curr=head

    print("\n##### BEFORE INSERT AT",k," POSITION #####")
    while curr!=None:
        print(curr.val,end=" -> ")
        curr=curr.next
    print("None")

    new_node=Node(new_node_value)

    if k==1:
        new_node.next=head
        head=new_node
        return

    curr=head
    for i in range(1,k-1):
        curr=curr.next

    new_node.next=curr.next
    curr.next=new_node

    curr=head
    print("************** AFTER INSERT AT",k,"th POSITION **************")
    while curr!=None:
        print(curr.val,end=" -> ")
        curr=curr.next
    print("None")
insert_at_place(25, 3)
```

```python
def insert_at_mid(new_node_value):
    global head
    curr=head
    new_node=Node(new_node_value)

    print("\n------ BEFORE INSERT AT MID ------")
    while curr!=None:
        print(curr.val,end=" -> ")
        curr=curr.next
    print("None")

    slow=head
    fast=head
    prev=None

    while fast!=None and fast.next!=None:
        prev=slow
        slow=slow.next
        fast=fast.next.next

    prev.next=new_node
    new_node.next=slow

    curr=head
    print("##### AFTER INSERT AT MID #####")
    while curr!=None:
        print(curr.val,end=" -> ")
        curr=curr.next
    print("None")
insert_at_mid(99)
```

```python

def delete_head_node():
    global head
    curr=head

    print("\n************** BEFORE DELETE HEAD **************")
    while curr!=None:
        print(curr.val,end=" -> ")
        curr=curr.next
    print("None")

    head=head.next

    curr=head
    print("------ AFTER DELETE HEAD ------")
    while curr!=None:
        print(curr.val,end=" -> ")
        curr=curr.next
    print("None")
delete_head_node()
```

```python
def delete_at_last():
    global head
    curr=head

    print("\n##### BEFORE DELETE LAST #####")
    while curr!=None:
        print(curr.val,end=" -> ")
        curr=curr.next
    print("None")

    curr=head
    while curr.next.next!=None:
        curr=curr.next

    curr.next=None

    curr=head
    print("************** AFTER DELETE LAST **************")
    while curr!=None:
        print(curr.val,end=" -> ")
        curr=curr.next
    print("None")
delete_at_last()
```

```python

def delete_at_position(k):
    global head
    curr=head

    print("\n------ BEFORE DELETE",k,"th POSITION NODE ------")
    while curr!=None:
        print(curr.val,end=" -> ")
        curr=curr.next
    print("None")

    if k==1:
        head=head.next
    else:
        curr=head
        for i in range(1,k-1):
            curr=curr.next
        curr.next=curr.next.next

    curr=head
    print("##### AFTER DELETE ",k,"th POSITION NODE #####")
    while curr!=None:
        print(curr.val,end=" -> ")
        curr=curr.next
    print("None")
delete_at_position(3)
```
